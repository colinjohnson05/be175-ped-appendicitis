```python
import pandas as pd
from data_pp import appendicitis_pp
from sklearn.pipeline import Pipeline
from sklearn.model_selection import StratifiedKFold
```

```python
data = appendicitis_pp('data/app_data.xlsx')
```

```python
# Extract classes and labels
groups = data['Class']
y = data['Diagnosis']

# Delete classes and labels from input data
X = data.drop(columns = ['Class', 'Diagnosis', 'Diagnosis_Presumptive'])

outer_cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
inner_cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)

for outer_fold, (outer_train_idx, test_idx) in enumerate(outer_cv.split(X, groups)):

    X_outer_train, X_test = X.iloc[outer_train_idx], X.iloc[test_idx]
    y_outer_train, y_test = y.iloc[outer_train_idx], y.iloc[test_idx]
    outer_groups          = groups.iloc[outer_train_idx]

    print(f"\nOuter Fold {outer_fold + 1}")
    print(f"  Train: {pd.Series(groups.iloc[outer_train_idx]).value_counts().to_dict()}")
    print(f"  Test:  {pd.Series(groups.iloc[test_idx]).value_counts().to_dict()}")

    for inner_fold, (inner_train_idx, val_idx) in enumerate(inner_cv.split(X_outer_train, outer_groups)):

        X_inner_train, X_val = X_outer_train.iloc[inner_train_idx], X_outer_train.iloc[val_idx]
        y_inner_train, y_val = y_outer_train.iloc[inner_train_idx], y_outer_train.iloc[val_idx]
        inner_groups         = outer_groups.iloc[inner_train_idx]

        print(f"  Inner Fold {inner_fold + 1}")
        print(f"    Train: {pd.Series(outer_groups.iloc[inner_train_idx]).value_counts().to_dict()}")
        print(f"    Val:   {pd.Series(outer_groups.iloc[val_idx]).value_counts().to_dict()}")
```
